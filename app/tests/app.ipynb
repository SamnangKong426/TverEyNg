# ESP32 Camera

```python
import cv2

url = "http://10.42.0.23/stream"


cap = cv2.VideoCapture(url)

if not cap.isOpened():
    print("Failed to connect to ESP32-CAM")
    exit()

while True:
    ret, frame = cap.read()  # Grab a frame
    if not ret:
        print("Failed to get frame")
        break

    cv2.imshow("ESP32 Live Feed", frame)  # Display frame

    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2 as cv
from ultralytics import YOLO

url = "http://10.42.0.23:81/stream"
model = YOLO("human.pt")
model.to("cpu")


cap = cv.VideoCapture(url)

if not cap.isOpened():
    print("Failed to connect to ESP32-CAM")
    exit()

while True:
    ret, frame = cap.read()  # Grab a frame
    if not ret:
        print("Failed to get frame")
        break

    results = model.predict(
        frame,
        conf=0.3,
        max_det=1,
        stream=True,
        
    )

    results = list(results)
    frame = results[0].plot()
    cv.imshow("ESP32 Live Feed", frame)  # Display frame

    # Press 'q' to exit
    if cv.waitKey(1) & 0xFF == ord("q"):
        break
```
