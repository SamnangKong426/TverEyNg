# ESP32 Camera

```python
import cv2

url = "http://192.168.1.5:81/stream"

cap = cv2.VideoCapture(url)

if not cap.isOpened():
    print("Failed to connect to ESP32-CAM")
    exit()

while True:
    ret, frame = cap.read()  # Grab a frame
    if not ret:
        print("Failed to get frame")
        break

    cv2.imshow("ESP32 Live Feed", frame)  # Display frame

    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
